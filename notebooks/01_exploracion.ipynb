```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt

RAIZ = Path.cwd()
if not (RAIZ / "src").exists():
    RAIZ = RAIZ.parent
sys.path.insert(0, str(RAIZ))
sys.path.insert(0, str(RAIZ / "src"))

from data import cargar_datos

velas, split = cargar_datos()
print(f"{len(velas)} velas | {velas.date.min().date()} → {velas.date.max().date()}")
velas[["date", "open", "high", "low", "close", "volume"]].tail()
```

```python
fig, ax = plt.subplots(figsize=(11, 3.5))
ax.plot(velas["date"], velas["close"], lw=0.8)
ax.set_yscale("log")
ax.set_title("Cierre BTC (escala log)")
fig.tight_layout()
plt.show()
```
